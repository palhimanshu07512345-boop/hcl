```python
"""Project 11: Public Library Circulation & Borrowing Analysis (tables only, no charts)."""
import sqlite3
import numpy as np
import pandas as pd

pd.set_option("display.width", 160)
pd.set_option("display.max_columns", 30)

def h(t):
    print("\n" + "=" * 70 + f"\n{t}\n" + "=" * 70)
```

```python
# ---- 1. Load & inspect
h("1. DIMENSIONS, COLUMNS, DTYPES")
df = pd.read_csv("11_library_circulation.csv")
print("Shape:", df.shape)
print(df.dtypes)
print(df.head())
```

```python
# ---- 2. Missing & duplicates
h("2. MISSING VALUES & DUPLICATE Loan_ID")
print(df.isna().sum())
dup_ids = df["Loan_ID"].duplicated().sum()
full_dups = df.duplicated().sum()
print(f"Duplicate Loan_ID rows (extra copies): {dup_ids}")
print(f"Fully identical duplicate rows: {full_dups}")
df = df.drop_duplicates(subset="Loan_ID", keep="first").reset_index(drop=True)
print("Shape after removing duplicate Loan_IDs:", df.shape)
```

```python
# ---- 3. Handle missing values
h("3. MISSING VALUE STRATEGY")
# Borrower_Age: impute with median age of the member's category (categories map to tight age bands)
n_age = df["Borrower_Age"].isna().sum()
df["Borrower_Age"] = df["Borrower_Age"].fillna(df.groupby("Member_Category")["Borrower_Age"].transform("median"))
print(f"Borrower_Age: {n_age} missing -> filled with the median age of the same Member_Category")
# Return_Date: cannot be invented. Treat as 'not returned / not recorded'; keep rows for borrow counts,
# exclude them from duration and overdue calculations.
df["Not_Returned"] = df["Return_Date"].isna()
print(f"Return_Date: {df['Not_Returned'].sum()} missing -> flagged Not_Returned=True; kept for borrow counts, "
      "excluded from duration/overdue/fine metrics (imputing would fabricate loan behaviour)")
```

```python
# ---- 4. Dates & loan duration
h("4. DATE CONVERSION & LOAN DURATION")
for c in ["Checkout_Date", "Due_Date", "Return_Date"]:
    df[c] = pd.to_datetime(df[c], dayfirst=True)
df["Loan_Duration"] = (df["Return_Date"] - df["Checkout_Date"]).dt.days
df["Days_Late"] = (df["Return_Date"] - df["Due_Date"]).dt.days.clip(lower=0)
df["Is_Overdue"] = (df["Return_Date"] > df["Due_Date"]).astype(float).where(~df["Not_Returned"])
print(df[["Checkout_Date", "Return_Date", "Loan_Duration"]].head())
print(df["Loan_Duration"].describe())

returned = df[~df["Not_Returned"]].copy()
returned["Is_Overdue"] = returned["Is_Overdue"].astype(bool)
```

```python
# ---- 5. Key metrics
h("5. KEY METRICS")
total = len(df)
avg_dur = returned["Loan_Duration"].mean()
overdue_pct = returned["Is_Overdue"].mean() * 100
print(f"Total books borrowed (unique loans): {total}")
print(f"Average loan duration (returned loans): {avg_dur:.2f} days")
print(f"Overall overdue percentage (returned loans): {overdue_pct:.2f}%")
```

```python
# ---- 6. Top 10 titles
h("6. TOP 10 MOST BORROWED TITLES")
top10 = df["Book_Title"].value_counts().head(10).rename_axis("Book_Title").reset_index(name="Borrow_Count")
top10.index += 1
print(top10)
```

```python
# ---- 7. Genre ranking
h("7. GENRE POPULARITY TABLE")
genre = df["Genre"].value_counts().rename_axis("Genre").reset_index(name="Borrow_Count")
genre["Share_%"] = (genre["Borrow_Count"] / total * 100).round(2)
genre["Rank"] = genre["Borrow_Count"].rank(ascending=False, method="min").astype(int)
print(genre)
```

```python
# ---- 8. Monthly volume
h("8. MONTHLY CIRCULATION TREND")
df["Month"] = df["Checkout_Date"].dt.to_period("M")
monthly = df.groupby("Month").size().rename("Borrow_Count").reset_index()
monthly["Month"] = monthly["Month"].astype(str)
print(monthly)
peak = monthly.loc[monthly["Borrow_Count"].idxmax()]
print(f"Peak reading month: {peak['Month']} with {peak['Borrow_Count']} loans")
```

```python
# ---- 9. Member categories
h("9. BORROWING FREQUENCY BY MEMBER CATEGORY")
cat = df["Member_Category"].value_counts().rename_axis("Member_Category").reset_index(name="Borrow_Count")
cat["Share_%"] = (cat["Borrow_Count"] / total * 100).round(2)
members = df.groupby("Member_Category")["Member_ID"].nunique().rename("Unique_Members")
cat = cat.merge(members, on="Member_Category")
cat["Loans_per_Member"] = (cat["Borrow_Count"] / cat["Unique_Members"]).round(2)
print(cat)
```

```python
# ---- 10. Overdue rates
h("10. OVERDUE RATE BY GENRE AND MEMBER CATEGORY")
od_genre = returned.groupby("Genre")["Is_Overdue"].agg(["count", "sum", "mean"])
od_genre.columns = ["Returned_Loans", "Overdue_Loans", "Overdue_Rate_%"]
od_genre["Overdue_Rate_%"] = (od_genre["Overdue_Rate_%"] * 100).round(2)
print(od_genre.sort_values("Overdue_Rate_%", ascending=False))
od_cat = returned.groupby("Member_Category")["Is_Overdue"].agg(["count", "sum", "mean"])
od_cat.columns = ["Returned_Loans", "Overdue_Loans", "Overdue_Rate_%"]
od_cat["Overdue_Rate_%"] = (od_cat["Overdue_Rate_%"] * 100).round(2)
print(od_cat.sort_values("Overdue_Rate_%", ascending=False))
print("\nOverdue rate (%) matrix: Genre x Member_Category")
print((returned.pivot_table(index="Genre", columns="Member_Category", values="Is_Overdue", aggfunc="mean") * 100).round(1))
```

```python
# ---- 11. Fines
h("11. ESTIMATED LATE FINES (assumed Rs 2 per day late)")
FINE_PER_DAY = 2
returned["Fine"] = returned["Days_Late"] * FINE_PER_DAY
print(f"Total estimated fines: Rs {returned['Fine'].sum():,.0f}")
print(f"Average fine per overdue loan: Rs {returned.loc[returned['Is_Overdue'], 'Fine'].mean():.2f}")
print("Fines by genre:")
print(returned.groupby("Genre")["Fine"].sum().sort_values(ascending=False))
print("Fines by member category:")
print(returned.groupby("Member_Category")["Fine"].sum().sort_values(ascending=False))
```

```python
# ---- 12. IQR outliers
h("12. UNUSUALLY LONG LOANS (IQR METHOD)")
d = returned["Loan_Duration"].to_numpy()
q1, q3 = np.percentile(d, [25, 75])
iqr = q3 - q1
lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
print(f"Q1={q1}, Q3={q3}, IQR={iqr}, lower fence={lower}, upper fence={upper}")
outliers = returned[returned["Loan_Duration"] > upper]
print(f"Unusually long loans: {len(outliers)} ({len(outliers)/len(returned)*100:.2f}% of returned loans)")
print("Longest 10 loans:")
print(outliers.nlargest(10, "Loan_Duration")[["Loan_ID", "Book_Title", "Genre", "Member_Category", "Loan_Duration"]])
print("Outliers by genre:")
print(outliers["Genre"].value_counts())
```

```python
# ---- 13. NumPy stats
h("13. NUMPY STATISTICS ON LOAN DURATION")
print(f"Mean   : {np.mean(d):.2f} days")
print(f"Median : {np.median(d):.2f} days")
print(f"Std Dev: {np.std(d, ddof=1):.2f} days (sample)")
```

```python
# ---- 14. Age groups
h("14. BORROWER AGE GROUPS")
bins = [0, 12, 17, 25, 40, 60, 120]
labels = ["Child (0-12)", "Teen (13-17)", "Young Adult (18-25)", "Adult (26-40)", "Middle Age (41-60)", "Senior (61+)"]
df["Age_Group"] = pd.cut(df["Borrower_Age"], bins=bins, labels=labels, right=True)
ag = df["Age_Group"].value_counts().sort_index().rename_axis("Age_Group").reset_index(name="Borrow_Count")
ag["Share_%"] = (ag["Borrow_Count"] / total * 100).round(2)
print(ag)
returned["Age_Group"] = df.loc[returned.index, "Age_Group"]
print("\nOverdue rate by age group (%):")
print((returned.groupby("Age_Group", observed=True)["Is_Overdue"].mean() * 100).round(2))
```

```python
# ---- Save outputs & database
h("SAVING OUTPUTS")
df.to_csv("library_cleaned.csv", index=False)
genre.to_csv("table_genre_popularity.csv", index=False)
monthly.to_csv("table_monthly_circulation.csv", index=False)
top10.to_csv("table_top10_books.csv")
od_genre.to_csv("table_overdue_by_genre.csv")
od_cat.to_csv("table_overdue_by_category.csv")

con = sqlite3.connect("library_circulation.db")
db = df.copy()
for c in ["Checkout_Date", "Due_Date", "Return_Date"]:
    db[c] = db[c].dt.strftime("%Y-%m-%d")
db["Month"] = db["Month"].astype(str)
db["Age_Group"] = db["Age_Group"].astype(str)
db.to_sql("loans", con, if_exists="replace", index=False)
con.execute("CREATE INDEX idx_genre ON loans(Genre)")
con.execute("CREATE INDEX idx_cat ON loans(Member_Category)")
con.commit()
print(pd.read_sql("SELECT Genre, COUNT(*) AS loans FROM loans GROUP BY Genre ORDER BY loans DESC LIMIT 3", con))
con.close()
print("Saved: library_cleaned.csv, library_circulation.db, table_*.csv")
```
